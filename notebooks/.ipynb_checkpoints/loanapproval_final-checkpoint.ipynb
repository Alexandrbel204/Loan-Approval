## Realization of Diffusion Generative Deep Learning Network

But firstly, I am going to preprocess data in some other way especially for NN

```python
X = df.drop(columns=['loan_status'])
y = df['loan_status']

categorical = ['person_gender', 'person_education', 'person_home_ownership', 'loan_intent', 'previous_loan_defaults_on_file']
numerical = [col for col in X.columns if col not in categorical]

# OneHot for categorical
enc = OneHotEncoder(sparse_output=False)
X_cat = enc.fit_transform(X[categorical])

# Scaling
scaler = MinMaxScaler()
X_num = scaler.fit_transform(X[numerical])

X_all = np.hstack([X_num, X_cat])
X_all = np.nan_to_num(X_all)

# Dividing into tensors
X_tensor = torch.tensor(X_all, dtype=torch.float32)
y_tensor = torch.tensor(y.values, dtype=torch.float32).unsqueeze(1)

# Only minoritary class
X_minority = X_tensor[y_tensor.squeeze() == 1]
```

```python
class DiffusionMLP(nn.Module):
    def __init__(self, input_dim, hidden_dim=256):
        super().__init__()
        self.input_dim = input_dim  # удобно для генерации
        # принимаем t как скаляр/вектор и конкатенируем (B,1)
        self.net = nn.Sequential(
            nn.Linear(input_dim + 1, hidden_dim),
            nn.SiLU(),
            nn.Linear(hidden_dim, hidden_dim),
            nn.SiLU(),
            nn.Linear(hidden_dim, input_dim)
        )

    def forward(self, x, t):
        """
        x: (B, D) or shapeable to (B, D)
        t: (B,1) or (B,) or scalar -> приведём к (B,1)
        Возвращаем (B, D)
        """
        # принудительно делаем x 2D: (B, D)
        x = x.reshape(x.shape[0], -1)
        # нормализуем/приводим t к (B,1)
        if t is None:
            # если t не передали — используем zeros
            t = torch.zeros((x.shape[0], 1), device=x.device, dtype=x.dtype)
        else:
            if t.dim() == 1:
                t = t.unsqueeze(1)
            elif t.dim() == 0:
                t = t.unsqueeze(0).unsqueeze(1).expand(x.shape[0], 1)
            # если t пришел как (B,1,1) — сожмём
            if t.dim() > 2:
                t = t.reshape(t.shape[0], -1)
                if t.shape[1] != 1:
                    t = t[:, :1]
        t = t.to(x.device).float()
        x_in = torch.cat([x, t], dim=1)  # (B, D+1)
        out = self.net(x_in)             # (B, D)
        out = out.reshape(x.shape[0], -1)  # 100% 2D
        return out
```

```python
model = DiffusionMLP(X_minority.shape[1])
optimizer = optim.Adam(model.parameters(), lr=1e-3)
loss_fn = nn.MSELoss()
```

```python
def p_sample_batch(model, x_t, t_int, betas, alpha_hat, alphas, device):
    """
    x_t: (B, D) или shapeable to (B,D)
    t_int: int scalar or tensor shape (B,)
    betas, alpha_hat, alphas: 1D tensors on device
    Возвращает x_{t-1} shape (B, D) (2D)
    """
    # ensure device
    device = device or ("cuda" if torch.cuda.is_available() else "cpu")
    x_t = x_t.to(device)
    # force 2D
    B = x_t.shape[0]
    x_t = x_t.reshape(B, -1)  # (B, D)
    D = x_t.shape[1]

    # build t tensor (B,)
    if isinstance(t_int, int):
        t = torch.full((B,), t_int, dtype=torch.long, device=device)
    else:
        t = t_int.to(device).long().reshape(-1)

    # get per-sample scalars
    beta_t = betas[t].unsqueeze(1)          # (B,1)
    a_hat_t = alpha_hat[t].unsqueeze(1)     # (B,1)
    a_t = alphas[t].unsqueeze(1)            # (B,1)

    # normalized time embedding used by model (we use t/T in [0,1])
    T = betas.shape[0]
    t_norm = (t.float() / float(T)).unsqueeze(1)  # (B,1)

    # predict noise -> ensure (B, D)
    pred_noise = model(x_t, t_norm)  # (B, D)

    # compute x0_pred and posterior mean
    sqrt_a_hat_t = torch.sqrt(a_hat_t)
    sqrt_1_minus_a_hat_t = torch.sqrt(1.0 - a_hat_t)

    x0_pred = (x_t - sqrt_1_minus_a_hat_t * pred_noise) / sqrt_a_hat_t  # (B, D)

    t_minus_1 = torch.clamp(t - 1, min=0)
    a_hat_t_minus_1 = alpha_hat[t_minus_1].unsqueeze(1)

    coef1 = (beta_t * torch.sqrt(a_hat_t_minus_1)) / (1.0 - a_hat_t)
    coef2 = ((1.0 - a_hat_t_minus_1) * torch.sqrt(a_t)) / (1.0 - a_hat_t)

    posterior_mean = coef1 * x0_pred + coef2 * x_t  # (B, D)

    posterior_variance = beta_t * (1.0 - a_hat_t_minus_1) / (1.0 - a_hat_t)  # (B,1)

    # sample noise only where t>0
    noise = torch.randn_like(x_t, device=device)
    mask = (t > 0).float().unsqueeze(1)
    x_prev = posterior_mean + mask * torch.sqrt(posterior_variance) * noise

    # ensure 2D
    x_prev = x_prev.reshape(B, D)
    return x_prev

# save generation
def generate_samples(model, n_samples=5000, betas=None, alpha_hat=None, alphas=None, device=None):
    """
    Возвращает numpy array shape (n_samples, D)
    model должен иметь attribute input_dim (см. класс выше) или мы возьмём последний Linear out_features
    """
    device = device or ("cuda" if torch.cuda.is_available() else "cpu")
    model = model.to(device)
    model.eval()

    assert betas is not None and alpha_hat is not None and alphas is not None, "Передайте betas, alpha_hat и alphas"

    T = betas.shape[0]
    # determine D robustly
    if hasattr(model, "input_dim"):
        D = model.input_dim
    else:
        last_lin = None
        for l in reversed(list(model.net)):
            if isinstance(l, nn.Linear):
                last_lin = l
                break
        if last_lin is None:
            raise RuntimeError("Не удалось определить выходной размер модели")
        D = last_lin.out_features

    # стартуем с нормального шума (B,D)
    x = torch.randn(n_samples, D, device=device)

    # обратный проход
    for t in range(T - 1, -1, -1):
        x = p_sample_batch(model, x, t, betas.to(device), alpha_hat.to(device), alphas.to(device), device)
        if x.ndim != 2:
            # приводим к 2D и продолжаем
            x = x.reshape(x.shape[0], -1)

    return x.detach().cpu().numpy()

def cosine_beta_schedule(timesteps, s=0.008):
    steps = timesteps + 1
    x = torch.linspace(0, timesteps, steps)
    alphas_cumprod = torch.cos(((x / timesteps) + s) / (1 + s) * math.pi / 2) ** 2
    alphas_cumprod = alphas_cumprod / alphas_cumprod[0]
    betas = 1 - (alphas_cumprod[1:] / alphas_cumprod[:-1])
    return torch.clip(betas, 0.0001, 0.02)


T = 300  

betas = cosine_beta_schedule(T)
alphas = 1.0 - betas
alpha_hat = torch.cumprod(alphas, dim=0)

def generate_samples_batched(model, total_samples, batch_size=1000, **kwargs):
    all_samples = []
    n_batches = (total_samples + batch_size - 1) // batch_size
    for i in range(n_batches):
        n = min(batch_size, total_samples - i * batch_size)
        batch = generate_samples(model, n_samples=n, **kwargs)
        if isinstance(batch, torch.Tensor):
            batch = batch.detach().cpu().numpy()
        all_samples.append(batch)
        torch.cuda.empty_cache()
    return np.concatenate(all_samples, axis=0)

# пример использования:
synthetic = generate_samples_batched(model, total_samples=10000, batch_size=1000,
                                     betas=betas, alpha_hat=alpha_hat, alphas=alphas, device="cuda" if torch.cuda.is_available() else "cpu")


print("Synthetic shape:", synthetic.shape)
```

```python
def sample_q(x0, t, noise, alpha_hat):
    device = x0.device
    alpha_hat = alpha_hat.to(device) 
    
    sqrt_alpha_hat = alpha_hat[t] ** 0.5
    sqrt_one_minus = (1 - alpha_hat[t]) ** 0.5
    return sqrt_alpha_hat.view(-1, 1) * x0 + sqrt_one_minus.view(-1, 1) * noise

n_epochs = 2000
for epoch in range(n_epochs):
    model.train()
    idx = torch.randint(0, X_minority.shape[0], (128,))
    x0 = X_minority[idx]
    device = model.net[0].weight.device

    t = torch.randint(0, T, (x0.shape[0],), device=device)
    noise = torch.randn_like(x0, device=device)
    x_t = sample_q(x0.to(device), t, noise, alpha_hat)


    pred_noise = model(x_t, t.float() / T)
    loss = loss_fn(pred_noise, noise)

    optimizer.zero_grad()
    loss.backward()
    optimizer.step()

    if epoch % 200 == 0:
        print(f"Epoch {epoch}/{n_epochs} | Loss: {loss.item():.4f}")
```

```python
num_dim = len(numerical)
cat_dim = X_cat.shape[1]

# Убедимся, что синтетические данные двумерные
if synthetic.ndim > 2:
    synthetic = synthetic.squeeze()

synthetic_num = synthetic[:, :num_dim]
synthetic_cat = synthetic[:, num_dim:num_dim + cat_dim]

# Обратное преобразование числовых признаков 
synthetic_num = scaler.inverse_transform(np.clip(synthetic_num, 0, 1))

# Обратное преобразование категориальных признаков 
synthetic_cat_bin = (synthetic_cat > 0.5).astype(int)

# делим на группы по исходным колонкам OneHotEncoder
cat_slices = []
start = 0
for cats in enc.categories_:
    end = start + len(cats)
    cat_slices.append((start, end))
    start = end

# гарантируем, что в каждой группе хотя бы один активный бит
for start, end in cat_slices:
    group = synthetic_cat_bin[:, start:end]
    all_zero = group.sum(axis=1) == 0
    if np.any(all_zero):
        # случайно активируем один бит в нулевых строках
        rand_idx = np.random.randint(0, group.shape[1], size=all_zero.sum())
        group[all_zero, rand_idx] = 1
    # обновляем группу обратно
    synthetic_cat_bin[:, start:end] = group

# теперь безопасно восстанавливаем категории
synthetic_cat = enc.inverse_transform(synthetic_cat_bin)

# Формируем датафрейм
synthetic_df = pd.DataFrame(synthetic_num, columns=numerical)
for i, col in enumerate(categorical):
    synthetic_df[col] = synthetic_cat[:, i]

# Добавляем таргет
synthetic_df['loan_status'] = 1

# Выводим информацию
print(f" Сгенерировано {len(synthetic_df)} строк")
print(f" Итоговая размерность: {df.shape[0] + len(synthetic_df)}")
```

```python
df_aug = pd.concat([df, synthetic_df], ignore_index=True)

# Again categorial features encoding 
df_encoded = pd.get_dummies(
    df_aug,
    columns=categorical,
    drop_first=True
)

# Again dividing 
X = df_encoded.drop(columns=['loan_status'])
y = df_encoded['loan_status']

X_train, X_test, y_train, y_test = train_test_split(
    X, y, stratify=y, test_size=0.2, random_state=42
)

# train boosting again
from xgboost import XGBClassifier
from sklearn.metrics import classification_report

xgb = XGBClassifier(
    n_estimators=300,
    max_depth=6,
    learning_rate=0.1,
    subsample=0.8,
    colsample_bytree=0.8,
    random_state=42
)
xgb.fit(X_train, y_train)
y_pred = xgb.predict(X_test)

print("\nClassification Report (with diffusion):")
print(classification_report(y_test, y_pred))
```
